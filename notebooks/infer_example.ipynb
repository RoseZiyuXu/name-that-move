# motion-clf — inference example


```python
from motion_clf import load_segments_batch, load_model, predict

TAG       = '20251023RoseMJDasha'   # must match what was used in save_artifacts
MODEL_DIR = './models'

# 1. Load test data — filename can be anything
X_test = load_segments_batch('./data/chunk_data-transposed_test_data.pkl')
print('Test data shape:', X_test.shape)   # (N, n_channels, n_timesteps)

# 2. Load feature extractor + trained learner
mrf, learn = load_model(model_dir=MODEL_DIR, tag=TAG)

# 3. Run inference
probas, preds = predict(X_test, mrf, learn)
print('Predictions:', preds)
print('Probabilities shape:', probas.shape)
```
